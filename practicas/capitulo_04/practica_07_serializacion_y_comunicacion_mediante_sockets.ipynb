```python
cliente, servidor = socket.socketpair()
cliente.settimeout(3)
servidor.settimeout(3)
with ThreadPoolExecutor(max_workers=1) as ejecutor:
    futuro = ejecutor.submit(servir, servidor)
    with cliente:
        cliente.sendall(json.dumps({"numeros": [4, 7, 1, 6]}).encode("utf-8") + b"\n")
        respuesta_bytes = b""
        while b"\n" not in respuesta_bytes:
            fragmento = cliente.recv(8)
            if not fragmento:
                raise ConnectionError("Respuesta incompleta")
            respuesta_bytes += fragmento
    futuro.result(timeout=4)
respuesta = json.loads(respuesta_bytes.decode("utf-8"))
print(respuesta)
```
